```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Create a sample dataset Age and Spending Score
data = {
    'Age': [20, 22, 25, 27, 30, 32, 45, 48, 50, 52],
    'Spending_Score': [80, 85, 78, 82, 75, 70, 30, 25, 20, 28]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Select the features for the model
X = df[['Age', 'Spending_Score']]
```

```python
# Create a KMeans model with 2 clusters
model = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)
model.fit(X)
```

```python
# Add the cluster labels to the original DataFrame(0 and 1)
df['Cluster'] = model.labels_

print(df)
```

```python
# Visualize the clusters using a scatter plot
plt.scatter(
    df['Age'],
    df['Spending_Score'],
    c=df['Cluster']
)

plt.xlabel('Age')
plt.ylabel('Spending Score')

plt.title('Customer Clustering using K-Means')

plt.grid()
plt.show()
```
