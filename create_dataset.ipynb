```python
import cv2
import os

# ---------------------------------------
# Dataset location
# ---------------------------------------

datasets = r"C:\face_datasets"
sub_data = "ashutosh"

# Complete path
path = os.path.join(datasets, sub_data)

# Create folders automatically
os.makedirs(path, exist_ok=True)

print("Dataset folder:")
print(path)

print("\nFolder exists:", os.path.exists(path))
```

```python
# Haar Cascade file
haar_file = r"C:\Users\ds512\OneDrive\Documents\open_cv\haarcascade_frontalface_default.xml"

# Check file
print("Haar Cascade exists:", os.path.exists(haar_file))

# Load Haar Cascade
face_cascade = cv2.CascadeClassifier(haar_file)

# Check whether it loaded correctly
if face_cascade.empty():
    print("ERROR: Haar Cascade could not be loaded.")
else:
    print("Haar Cascade loaded successfully!")
```

```python
# ---------------------------------------
# Open webcam
# ---------------------------------------

webcam = cv2.VideoCapture(0)

if not webcam.isOpened():
    print("ERROR: Could not open webcam.")
else:

    print("Camera started successfully!")
    print("Look at the camera.")
    print("Press ESC to stop.")

    # Image size
    width = 130
    height = 100

    # Starting image number
    count = 1

    # ---------------------------------------
    # Capture images
    # ---------------------------------------

    while count <= 30:

        # Read webcam frame
        ret, frame = webcam.read()

        if not ret:
            print("ERROR: Could not read camera.")
            break

        # Convert to grayscale
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # Detect faces
        faces = face_cascade.detectMultiScale(
            gray,
            scaleFactor=1.3,
            minNeighbors=5
        )

        # ---------------------------------------
        # Process detected faces
        # ---------------------------------------

        for (x, y, w, h) in faces:

            # Draw rectangle around face
            cv2.rectangle(
                frame,
                (x, y),
                (x + w, y + h),
                (255, 0, 0),
                2
            )

            # Crop face
            face = gray[y:y + h, x:x + w]

            # Resize face
            face_resize = cv2.resize(
                face,
                (width, height)
            )

            # File name
            filename = os.path.join(
                path,
                f"{count}.png"
            )

            # Save image
            success = cv2.imwrite(
                filename,
                face_resize
            )

            if success:
                print(f"Image {count} saved")
                count += 1
            else:
                print(f"ERROR: Could not save image {count}")

            # Only capture one face
            break

        # ---------------------------------------
        # Show webcam
        # ---------------------------------------

        cv2.imshow(
            "Face Capture - Press ESC to Stop",
            frame
        )

        # Wait for keyboard
        key = cv2.waitKey(10) & 0xFF

        # ESC key
        if key == 27:
            print("Capture stopped by user.")
            break

    # ---------------------------------------
    # Close webcam
    # ---------------------------------------

    webcam.release()
    cv2.destroyAllWindows()

    print("\n--------------------------------")
    print("Face capture completed!")
    print("--------------------------------")
    print("Images saved in:")
    print(path)
```
