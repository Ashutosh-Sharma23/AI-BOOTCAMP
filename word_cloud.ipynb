```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter
```

```python
text = """
My hobby is playing cricket. Cricket is one of my favorite sports, and I enjoy playing it whenever I get free time. I started playing cricket at a young age with my friends, and gradually it became an important part of my daily life. I like cricket because it is not only an entertaining game but also teaches many important lessons about life. When I play cricket, I feel energetic, active, and relaxed. It helps me stay physically fit and improves my concentration and decision-making skills.

I usually play cricket with my friends in the evening or whenever we get an opportunity. I enjoy both batting and bowling, but I especially like batting because it requires good concentration, timing, and patience. Playing cricket has also taught me the importance of teamwork. In a cricket match, every player has a different role, and the team can succeed only when everyone works together. It teaches me how to handle both success and failure positively.

Cricket has also helped me develop qualities such as discipline, leadership, patience, and confidence. Sometimes we win a match, and sometimes we lose, but every match gives me an opportunity to learn something new. I also enjoy watching professional cricket matches and observing how experienced players handle pressure and make decisions.

In the future, I want to continue playing cricket because it keeps me healthy and helps me maintain a positive mindset. For me, cricket is not just a hobby; it is a way to stay active, learn teamwork, reduce stress, and enjoy my free time. the photographer's perspective and artistic style. I like Photography Photography Photography Photography Photography Photography Photography Photography .
"""
print(text)
```

```python
# Tokenization of words
words = word_tokenize(text.lower())  # Convert to lowercase for uniformity
print(words)
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
#  Create a WordCloud
word_freq = Counter(filtered_words) 
print(word_freq)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
```

```python
# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.show()
```

```python
plt.imsave('wordcloud.png', wordcloud)  # Save the word cloud image
```
