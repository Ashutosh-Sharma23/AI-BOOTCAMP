```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Student_ID": [101, 102, 103, 104,105,106,107,108],
    "Name": ["Amit", "Neha", "Rahul", "Priya","Karan","Anjali","Ravi","Sneha"],
    "Department": ["CSE", "CSE", "ECE", "ECE","CSE","IT","IT","CSE"],
    "Gender": ["Male", "Female", "Male", "Female","Male","Female","Male","Female"],
    "Marks": [85, 92, 78, None,88,95,70,90],
    "Attendance": [90, 95, 80, 85,None,98,75,92],
    "Age": [20, 21, 20, 22,21,20,22,21],
    "Remarks": ["Good", "Excellent", "Good", "Average","Good","Excellent","Average","Excellent"]

}
```

```python
# create a DataFrame from the data dictionary
df = pd.DataFrame(data)

# create csv file from the DataFrame
df.to_csv('student_data.csv', index=False) 

# Read the CSV file into a new DataFrame
df_new = pd.read_csv('student_data.csv')
df_new
```

```python
# find null values in the DataFrame
print(df_new.isnull().sum())
```

```python
# Relpace null values in the 'Marks' column with the mean of the column
df_new['Marks'] = df_new['Marks'].mean()
# Replace null values in the 'attendance' column with the mean of the column
df_new['Attendance'] = df_new['Attendance'].mean()
df_new
```

```python
# Create Sine wave using numpy and matplotlib
fig , ax = plt.subplots(figsize=(10, 5))
radius = np.linspace(0, 2 * np.pi, 100)
sine_wave = np.sin(radius)
ax.plot(radius, sine_wave)
```
